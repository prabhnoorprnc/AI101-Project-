```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("../data/Sample-Data-CPU-Performance.csv")
```

```python
display(df.head())
print(df.shape)
df.info()
```

```python
quantity_cols = [
    'CycleTime',
    'MinimumMainMemory',
    'MaximumMainMemory',
    'CacheSize'
]

for col in quantity_cols:
    df[col] = df[col].str.extract(r'Quantity\[\s*(-?\d+(?:\.\d+)?)')[0]
    df[col] = pd.to_numeric(df[col])
```

```python
print("Missing values:")
print(df.isnull().sum())
```

```python
print("Duplicate rows:", df.duplicated().sum())

df = df.drop_duplicates()
```

```python
features = [
    'CycleTime',
    'MinimumMainMemory',
    'MaximumMainMemory',
    'CacheSize',
    'MinimumNumberOfChannels',
    'MaximumNumberOfChannels'
]

target = 'EstimatedPerformance'

X = df[features]
y = df[target]

print("\nCleaned dataset:")
print(X.head())

print("\nTarget variable:")
print(y.head())

print("\nFinal dataset shape:", df.shape)
```
